##3.1创建dataframe对象

```python
##1.字典+列表
#1.准备数据集
import pandas as pd
info = {
    'name':['Tom','Felix','Harry','Teresa'],
    'gender':['male','male','male','female']
}
#2.封装数据集
df1 = pd.DataFrame(data = info)

#3打印结果
print(df1)
```

```python
##2列表+元组
import pandas as pd
info = [('Tom','male',39)
        ,('Felix','male',29),('Harry','male',19),('Teresa','female',29)]
df2 = pd.DataFrame(data = info,columns = ['name','gender','age'])
print(df2)
```

```python
##3.通过numpy的ndarray创建DataFrame
import pandas as pd
import numpy as np
arr1 = np.arange(12).reshape(3,4)
df3 = pd.DataFrame(data =arr1,columns = ['A','B','C','D'])
print(df3)
```

##3.2 DataFrame 简单案例

```python
#1生成10名同学5门课程的成绩数据，要求成绩在0-100之间
import pandas as pd
import numpy as np
score_df =  pd.DataFrame(data = np.random.randint(0,101, size=(10,5))) #包左不包右
#2修改DataFrame对象的列名和索引列值
column_names = ['语文','数学','英语','物理','化学']
index_names = ['student'+ str(i) for i in range(score_df.shape[0])] #df.shape 返回 (行数, 列数)，shape[0] = 当前表格总行数。 
#3具体修改DataFrame对象 列名和索引值
score_df.columns = column_names
score_df.index = index_names
score_df.rename(columns = {'语文':'math', '数学':'Chinese'},index = {'student0':'同学0','student1':'同学1'},inplace = True) #rename精准修改
#4打印结果
print(score_df)
```

##3.4DaraFrame 对象方法（基本函数）

```python
# head:显示前n行数据,默认显示前5行数据
print(score_df.head())
```

```python
#tail 显示后n行数据，默认显示后5行数据
print(score_df.tail(2))
```

```python
#info():显示DataFrame对象的基本信息
print(score_df.info())
```

```python
#describe():显示DataFrame对象的数值型数据的统计信息
print(score_df.describe())
```

```python
#修改行列索引值
score_df.rename(columns={'math':'语文','Chinese':'数学'},index = {'同学0':'student0','同学1':'student1'},inplace = True) #rename精准修改
print(score_df)

stu = ['stu'+ str(i) for i in range(score_df.shape[0])] #df.shape 返回 (行数, 列数)，shape[0] = 当前表格总行数。
score_df.index = stu
print(score_df)
```

```python
#reset_index()重设索引,默认drop =False,保留原索引列
score_df.reset_index(drop = True,inplace = True) #drop = False,保留原索引列
print(score_df)
```

```python
#set_index()设置新的索引，默认drop =True,删除原索引列
score_df.set_index(['语文','数学'],drop = True,inplace = True) #drop = False,保留原索引列
print(score_df)
```

##3.5 pandas的数据类型
object：等于python的string
category：分类类型，针对分类数据操作更快
nan：空值

```python
#1创建一个DataFrame对象
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'name':['Tom','Felix','Harry','Teresa'],
    'gender':['male','male',np.nan,'classify'],
    'age':[39,29,19,29]
})
df.info()
```

```python
#2datetime
df2 = pd.DataFrame({'date': ['2021-01-01','2021-01-02','2021-01-03','2021-01-04']},dtype = 'datetime64[ns]')
print(df2)
```

```python
#3timedelta 日期差类型
st_date = pd.to_datetime('2010-01-01')
end_date = pd.to_datetime('2010-1-31')
print(type(end_date - st_date))
```

```python
#4category
s1 = pd.Series(['male','female','classify'], dtype='category')
s2 = pd.Series(['male','female','classify'])
print(s1.dtype)
print(s2.dtype)
```
